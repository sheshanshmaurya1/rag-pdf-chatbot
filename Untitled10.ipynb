```python
import os

from google.colab import userdata

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_openai import OpenAIEmbeddings
from langchain_openai import ChatOpenAI

from langchain_community.vectorstores import Chroma

from langchain_core.prompts import ChatPromptTemplate
```

```python
OPENAI_API_KEY = userdata.get("OPENAI_API_KEY").strip()

os.environ["OPENAI_API_KEY"] = OPENAI_API_KEY
```

```python
DOC_PATH = "/content/project-report-Sheshansh mauryapdf.pdf"

CHROMA_PATH = "/content/chroma_db"
```

```python
loader = PyPDFLoader(DOC_PATH)

pages = loader.load()

print(len(pages))
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(pages)

print(len(chunks))
```

```python
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small",
    openai_api_key=OPENAI_API_KEY
)
```

```python
db = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory=CHROMA_PATH
)
```

```python
query = "Give only name person how have submitted this report and project is around which topic?"
```

```python
docs = db.similarity_search_with_score(
    query,
    k=5
)
```

```python
context_text = "\n\n".join(
    [doc.page_content for doc, score in docs]
)
```

```python
PROMPT_TEMPLATE = """
Answer the question using ONLY the context below.

Context:
{context}

Question:
{question}

Provide a detailed answer.

Do not invent information.

Do not mention the context itself.
"""
```

```python
prompt = ChatPromptTemplate.from_template(
    PROMPT_TEMPLATE
)

final_prompt = prompt.format(
    context=context_text,
    question=query
)
```

```python
llm = ChatOpenAI(
    model="gpt-5.5",
    temperature=0
)
```

```python
response = llm.invoke(final_prompt)

print(response.content)
```
